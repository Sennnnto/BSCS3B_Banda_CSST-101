Numeric Operation

```r
x <- 5
y <- 7

sum <- x + y
diff <- x - y
quo <- x / y
mul <- x * y

print(paste0("The sum is ",sum))
print(paste0("The difference is ",diff))
print(paste0("The quotient is ",quo))
print(paste0("The product is ",mul))

cat("\nThe Sum is", sum ,"\n")
cat("The Difference is", diff ,"\n")
cat("The Product is", mul ,"\n")
cat("The Quotient is", quo ,"\n")
```
